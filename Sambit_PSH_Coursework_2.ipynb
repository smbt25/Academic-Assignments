**Interpretation (EMH)**

Under the semi-strong form of Efficient Market Hypothesis (EMH), publicly available information should be quickly incorporated into prices quickly, and technical and fundamental analysis cannot deliver abnormal returns consistently after adjusting for risk and costs.

In our case:
- The very small AR at (t = 0) shows **no large, clean announcement shock**. This means that Pershing's stock underperformed immediately compared to market when the event was reported on Nov 14, 2024. This means that investors of Pershing were skeptical of its investment in Nike. At that time Nike was performing poorly experiencing revenue and profit margin decline. Such underperformance of Pershing's price is a short term reaction by investors and is immediate and extremely small and no different than 0. The continued negative abnormal return till t=+6 along with decreasing cumulative abnormal return shows a modest post event drift which is not statistically different from 0 and can be concluded as noise from regular trading of Pershing on the stock exchange.

- The positive Cumulative Absolute Return is modest at (t = 0). This is consistent with a modest alpha generation that Pershing follows as it is an actively managed fund. However there is not much fluctuation due to the event and it looks more like noisy adjustment and overlapping news than a stable trading opportunity as any abnormal return is quickly arbitraged away.

Thus there is **no significant post-event drift** and is indicative of semi-strong form of market efficiency since no statistically significant or clear abnormal movement is observed around the event showing indications of any post event drift.


```python
# MARKET MODEL TEST
# estimating alpha and beta on a pre-event window
# downloading 200 days of trading data ending 20 days before to avoid contamination

est_start = (event_ts - pd.Timedelta(days=220)).strftime("%Y-%m-%d")    # creating start date
est_end   = (event_ts - pd.Timedelta(days=20)).strftime("%Y-%m-%d")     # creating end date

p_est= yf.download([co, ft],
                start= est_start,
                end= est_end,
                auto_adjust= True,
                progress= False
                )["Close"]

px_est= p_est.rename(columns={co: "PSH", ft: "FTSE"})                  # renaming columns
ret_est= px_est.pct_change().dropna()                                  # calcualting percentage returns

beta_mm = ret_est["PSH"].cov(ret_est["FTSE"]) / ret_est["FTSE"].var()  # calculating beta on the estimate
alpha_mm = ret_est["PSH"].mean() - beta_mm * ret_est["FTSE"].mean()    # calculating alpha

print ("The left day for the pre event window is   ",est_start)
print ("The right day for the pre event window is  ", est_end)
print("")
print("The beta during the pre event window is   ", round(beta_mm,4))
print("The alpha during the pre event window is ", round(alpha_mm,4))
print("")
print("")

#---------------------------------------------------------------------------

# applying alpha and beta to the pre event estimation window

exp_event= alpha_mm + beta_mm*win["FTSE"]

win["Ab_Ret_mm"]= win["PSH"] - exp_event
win["Cum_AR_mm"]= win["Ab_Ret_mm"].cumsum()

ar0_mm = float(win.loc[win["Date_Diff"] == 0, "Ab_Ret_mm"])
car_tot_mm = float(win["Cum_AR_mm"].iloc[-1])

print("Abnormal Return of pre event at t=0 (%) =", round(100 * ar0_mm, 2))
print("Cumulative Abnormal Return of pre event in (%) =", round(100 * car_tot_mm, 2))
print("")
print("")

#---------------------------------------------------------------------------

# Conducting t-stats to check for noise

eps_est= ret_est["PSH"] - (alpha_mm + beta_mm * ret_est["FTSE"])
sigma_eps = eps_est.std(ddof=1)

n_win = len(win)                                          # counting no of tradind days in our window

# computing t stat for abnormal return and cumulative abnormal return. if |t|>2, it is significant at 5% level.

t_ar0_mm = ar0_mm /  sigma_eps
t_car_tot_mm = car_tot_mm / (sigma_eps * (n_win**0.5))

print("t-stat (MM) Abnormal Return at t=0 =", round(t_ar0_mm, 2))
print("t-stat (MM) Cumulative Abnormal Return (t= -" +str(pre_days), "to t=" +str(post_days),") =", round(t_car_tot_mm, 2))
print("")
print("")
```

Here we use the **Market Model** to refine our earlier market adjusted returns which is to give us a more a accurate measure of abnormal return.

The estimated alpha and beta to the event window to get the expected abnormal return.

Abnormal return as per Market model is -0.3% means Pershing slightly underperformed what the market model predicted. It is a small, normal fluctuation and not a strong reaction as it is not statistically different than 0.

Cumulative Abnormal Return as per market model over 21 day window is (t=-10 to t= +10) is +0.82% meaning Pershing slightly outperformed the market.

Together these results show no evidence of abnormal, market-moving announcements and there is no indication of any post event drift. This is consistent with semi-strong form of market efficiency where the information was captured quickly into the prices and residual market movements are just normal noise from day to day trading activities.

If |t|> 2, it means that the event was statistically significant.
Here the t stat value of AR_mm (abnormal return of market model at t=0) is -0.23% and t stat value of CAR_mm (cumulative abnormal return of market model) is 0.82%. Both the values are close to 0 at 5% significance means these are **not significant**.

Hence the event did not materially affect the prices of Pershing Square Holdings.

**3B>**

As per the Modigliani Miller (MM) theorem the market value of a company is the present value of the future earnings and is independent of its capial structure in perfect market with no taxes, transaction costs and symmetric information. However in real world, capital structure can affect the value of a firm.

In **cash financed merger**, the acquirer bears all the risk and information problems. The target shareholders exit completely so the target bears little to no risk. If the Acquirer overpays due to bad information or if the synergies do not get realised, the entire loss is borne by the acquirer. Cash financed merger generally leads to **value creation** for the combined entity as the net debt increases which increases tax shield on interest payments.

In **stock financed merger**, the target shareholders also become the shareholders in the combined entity. Thus both acquirer and target shareholders both share the risk and rewards. Stock financed merger leads to **value redistribution** as the both acquirer and target shareholders share risk and reward from the combined entity and thus leads to dilution of the value.

**3C>**

If the acquirer share are undervalued by ∆.

In **Cash Merger** existing shareholders hold undervalued shares and will gain when the mispricing corrects itself. Target shareholders have already received premium and are no longer entitled to any future gains. **Only Acquirer shareholders will gain ∆ when the mispricing corrects in a cash merger.** Target shareholders will lose out on the mispricing gains.

In **Stock Merger** since target shareholders are also getting the shares of the combined entity. Hence both the acquirer and targt shareholder are both entitled to the gain from mispricing. Thus **both acquirer and target shareholders will share ∆ gains when the mispricing corrects in stock merger.** This is considered as a loss by managers of the acquiring company as the shareholders of the acquiring company lose out on the entire benefit of the mispricing gains.

If the acquirer's share are undervalued perceived misvaluation due to asymmetric information, managers will likely **chose cash merger for financing the merger as corrections from mispricing will accrue only to existing shareholders** as managers act in interests of the current owners and paying in stock will lead them to issue undervalued stock to target shares which can be considered as *value destroying* for the current owners and would want to avoid.

**3D>**

The total merger gain is the present value of the after tax synergies of the combined entity. The gains is divided between the acquirer and target shareholders on the basis of premium and integrations costs paid for the merger.

The shareholder of the target company gets paid as the negotiated premium, thus shareholders of the target company receive no gain in a cash merger. The amount of premium may include compensation that the shareholder of the target company needs as share of the synergies that will be acheived by the combined entity. Histprocally, target shareholders always gain from mergers. However, they gain more in a cash merger than a stock merger as stock merger often indicates that the standalone value of the acquirer or the expected synergies are low.

The shareholders of the acquiring company gain from the merger only if the present value of the after tax synergies is more than the premium paid and integration costs for the merger. This means that synergies are accurately estimated and timely realised. The merger should lead to synergies like economies of scale, economies of scope, vertical integration and taking advantage of complementary of resources. If the synergies are effectively realised in a timely manner then the acquirer's shareholders will benefit from the combined entity. Acquiring shareholder gain higher in a cash financed merger when the acquirer is convinced of high synergies due to asymmetric information or if managers believe that the acquirer's standalone value is undervalued.
